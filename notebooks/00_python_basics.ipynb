## 5. if (ถ้า...ก็...)
ใช้ตัดสิน  **เยื้อง 4 ช่อง**ใต้ `if` สำคัญมาก เป็นวิธีที่ Python รู้ว่าบรรทัดไหนอยู่ในเงื่อนไข

````python
found = 4
expect = 5
if found == expect:
    verdict = 'OK'
elif found < expect:
    verdict = 'NG ขาด'
else:
    verdict = 'NG เกิน'
print(verdict)
````

**แบบฝึก:** เขียนฟังก์ชันตัดสินให้ครบ 3 กรณี  (`==` เท่ากับ, `<` น้อยกว่า, `>` มากกว่า)

````python
def judge(found, expect):
    if found == expect:
        return 'OK'
    return 'NG'                # TODO: แยกเป็น 'NG ขาด' เมื่อน้อยกว่า และ 'NG เกิน' เมื่อมากกว่า

check('5a. ครบ', judge(5, 5), 'OK')
check('5b. ขาด', judge(4, 5), 'NG ขาด')
check('5c. เกิน', judge(6, 5), 'NG เกิน')
````
